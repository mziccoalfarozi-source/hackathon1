```python
import os
from pathlib import Path
import pyreadr
import pandas as pd

base_dataset_dir = Path("../Dataset/Hospital Triage and Patient History")
rdata_path = base_dataset_dir / "5v_cleandf.rdata"
csv_output_path = base_dataset_dir / "hospital_triage_clean.csv"

if not rdata_path.exists():
    raise FileNotFoundError(f"File RData tidak ditemukan: {rdata_path.resolve()}")

print("Sedang membaca file .rdata, harap tunggu...")
result = pyreadr.read_r(str(rdata_path))

if len(result) == 0:
    raise ValueError("Objek dalam file .rdata kosong atau gagal terbaca.")

df_hospital = result[list(result.keys())[0]]
print(f"Berhasil memuat data! Total: {df_hospital.shape[0]} baris pasien dan {df_hospital.shape[1]} kolom.")

df_hospital.to_csv(csv_output_path, index=False)
print(f"CSV berhasil disimpan di: {csv_output_path.resolve()}")

if csv_output_path.exists() and csv_output_path.stat().st_size > 0:
    print("Validasi CSV: file ada dan tidak kosong.")
else:
    raise RuntimeError("CSV gagal dibuat atau file kosong.")

rdata_path.unlink()
print(f"File RData dihapus untuk menghemat ruang: {rdata_path.resolve()}")
```

```python
from pathlib import Path
import pandas as pd

hospital_csv_path = Path("../Dataset/Hospital Triage and Patient History/hospital_triage_clean.csv")

# Menangani perbedaan penamaan folder/file dataset sintetis di workspace
synthetic_candidates = [
    Path("../Dataset/Synthetic Medical Triage/synthetic_medical_triage_priority_dataset.csv"),
    Path("../Dataset/Synthetic Medical Triage Priority Dataset/synthetic_medical_triage.csv"),
]
synthetic_csv_path = next((p for p in synthetic_candidates if p.exists()), None)

if synthetic_csv_path is None:
    raise FileNotFoundError(
        "Dataset sintetis tidak ditemukan di path kandidat manapun. "
        "Periksa nama folder/file pada direktori Dataset."
    )

if not hospital_csv_path.exists():
    raise FileNotFoundError(f"Dataset hospital tidak ditemukan: {hospital_csv_path.resolve()}")

df_synthetic = pd.read_csv(synthetic_csv_path)
df_hospital = pd.read_csv(hospital_csv_path)

print(f"Synthetic dataset loaded dari: {synthetic_csv_path.resolve()}")
print(f"Hospital dataset loaded dari: {hospital_csv_path.resolve()}")

print("\n=== Synthetic Dataset: .info() ===")
df_synthetic.info()
print("\n=== Synthetic Dataset: .head() ===")
display(df_synthetic.head())

print("\n=== Hospital Dataset: .info() ===")
df_hospital.info()
print("\n=== Hospital Dataset: .head() ===")
display(df_hospital.head())
```

```python
# Ringkasan cepat untuk persiapan merging
common_columns = sorted(set(df_synthetic.columns).intersection(df_hospital.columns))
print(f"Jumlah kolom synthetic: {len(df_synthetic.columns)}")
print(f"Jumlah kolom hospital: {len(df_hospital.columns)}")
print(f"Kolom yang sama persis: {len(common_columns)} -> {common_columns}")

keyword_candidates = [
    c for c in df_hospital.columns
    if any(k in c.lower() for k in ["heart", "blood", "pressure", "oxygen", "temp", "triage", "esi", "pain", "arrival", "visit", "chronic"])
]
print(f"Jumlah kandidat kolom hospital (berdasarkan keyword klinis): {len(keyword_candidates)}")
print("Contoh 30 kolom kandidat pertama:")
print(keyword_candidates[:30])
```
